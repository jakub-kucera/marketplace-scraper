```python
import requests

proxy = {
    'http':  'socks5://127.0.0.1:9050',
    'https': 'socks5://127.0.0.1:9050',
}

url = 'https://icanhazip.com'
# url = 'http://httpbin.org/ip'

r = requests.get(url, proxies=proxy)  # using TOR network
print('   Tor IP:', r.text)

r = requests.get(url)
print('direct IP:', r.text)
```

```python
from time import sleep

import requests
from stem import Signal
from stem.control import Controller


def get_current_ip():
    session = requests.session()

    # TO Request URL with SOCKS over TOR
    session.proxies = {
        'http': 'socks5h://localhost:9050',
        'https': 'socks5h://localhost:9050'
    }
    try:
        # r = session.get('http://httpbin.org/ip')
        r = session.get('https://icanhazip.com')
    except Exception as e:
        print (str(e))
    else:
        return r.text


def renew_tor_ip():
    with Controller.from_port(port = 9051) as controller:
        controller.authenticate(password="root")
        controller.signal(Signal.NEWNYM)


def renew_ip_and_check():
    old_ip = get_current_ip()
    print(f"Old IP: {old_ip}")
    renew_tor_ip()
    while True:
        new_ip = get_current_ip()
        if old_ip != new_ip:
            print (f"New IP: {new_ip}")
            break
        sleep(1)

    print("IP renewed")
    
def bazos_status():
        # r = requests.get("https://www.bazos.cz", proxies={
    r = requests.get("https://pc.bazos.cz/monitor/", proxies={
        'http': 'socks5h://localhost:9050',
        'https': 'socks5h://localhost:9050'
    })
    print(r.text)
    return r.status_code
    

# renew_ip_and_check()
bazos_status()
```
